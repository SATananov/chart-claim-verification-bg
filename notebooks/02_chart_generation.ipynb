# Chart Generation

This notebook will create chart images from clean numerical tables.

```python
import matplotlib.pyplot as plt
import pandas as pd

example = pd.DataFrame(
    {
        "year": [2019, 2020, 2021, 2022, 2023],
        "value": [5.1, 5.8, 5.4, 4.9, 4.6],
    }
)

example
```

```python
plt.figure(figsize=(8, 4))
plt.plot(example["year"], example["value"], marker="o")
plt.title("Example Statistical Chart")
plt.xlabel("Year")
plt.ylabel("Value")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

The example is only used to test the chart code. Final charts will be created from the real dataset.
