# Chart Generation

In this notebook, I will turn the cleaned numerical data into a chart image.

This is the first computer vision preparation step because the image will later become an input to a CNN model.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import pandas as pd

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.chart_generation import save_line_chart
```

```python
data_path = (
    project_root
    / "data"
    / "processed"
    / "eu_unemployment_2019_2025.csv"
)

data = pd.read_csv(data_path)
data
```

## Preview

The chart is first displayed inside the notebook so I can check the title, labels and values.

```python
plt.figure(figsize=(8, 4))

plt.plot(
    data["year"],
    data["unemployment_rate"],
    marker="o",
)

plt.title("EU Unemployment Rate, 2019-2025")
plt.xlabel("Year")
plt.ylabel("Unemployment rate (%)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Save the chart

The image is saved as a PNG file. Later, many chart images will be generated for model training.

```python
output_path = (
    project_root
    / "data"
    / "generated"
    / "charts"
    / "eu_unemployment_2019_2025.png"
)

saved_path = save_line_chart(
    data=data,
    output_path=output_path,
    title="EU Unemployment Rate, 2019-2025",
)

print("Saved chart:", saved_path)
print("File exists:", saved_path.exists())
print("File size:", saved_path.stat().st_size, "bytes")
```

## Result

The numerical data can now be converted into a consistent image file.

This is only the first chart. The next chart-generation step will create more examples and controlled visual variations.
