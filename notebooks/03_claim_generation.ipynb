# Claim Generation

In this notebook, I will generate short text claims from the yearly
unemployment values.

Each claim will receive one of three labels:

- supported
- refuted
- not enough information

## Main idea

For each pair of adjacent years, I will create:

1. one correct comparison;
2. one incorrect comparison;
3. one comparison that uses a year outside the chart.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import pandas as pd

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.claims import generate_claim_dataset
```

```python
data_path = (
    project_root
    / "data"
    / "processed"
    / "eu_unemployment_2019_2025.csv"
)

data = pd.read_csv(data_path)
data
```

```python
claims = generate_claim_dataset(data)

claims.head(9)
```

## Label balance

A balanced first sample makes it easier to check the generation logic
before a larger dataset is created.

```python
label_counts = (
    claims["label"]
    .value_counts()
    .reindex(
        [
            "supported",
            "refuted",
            "not_enough_information",
        ]
    )
)

label_counts
```

```python
plt.figure(figsize=(7, 4))

label_counts.plot(kind="bar")

plt.title("Generated Claims by Label")
plt.xlabel("Label")
plt.ylabel("Number of claims")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
output_path = (
    project_root
    / "data"
    / "processed"
    / "eu_unemployment_claims.csv"
)

claims.to_csv(
    output_path,
    index=False,
)

print("Saved claims:", output_path)
print("Rows:", len(claims))
print("Missing values:", claims.isna().sum().sum())
```

## Result

The first claim dataset contains 18 rows and the three labels are
balanced.

These claims are generated from one chart only. Later, the same logic
will be applied to many charts so the neural network has enough training
