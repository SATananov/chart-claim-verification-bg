# Data Understanding

In this notebook, I will load a small real dataset and check its basic
structure before using it in later steps.

## Dataset

The dataset contains the annual unemployment rate for the European Union
from 2019 to 2025.

The values are percentages of the labour force aged 15 to 74.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

data_path = (
    project_root
    / "data"
    / "processed"
    / "eu_unemployment_2019_2025.csv"
)

data = pd.read_csv(data_path)
data
```

```python
print("Rows:", len(data))
print("Columns:", list(data.columns))
print()
print(data.dtypes)
```

```python
print("Missing values:")
print(data.isna().sum())
print()
print("Duplicate rows:", data.duplicated().sum())
```

```python
data.describe()
```

## Visualization

A line chart is suitable because the values are ordered by year.

```python
plt.figure(figsize=(8, 4))

plt.plot(
    data["year"],
    data["unemployment_rate"],
    marker="o",
)

plt.title("EU Unemployment Rate, 2019-2025")
plt.xlabel("Year")
plt.ylabel("Unemployment rate (%)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## First observations

The rate increased in 2020 and then decreased in the following years.
The 2025 value is slightly higher than the 2024 value.
