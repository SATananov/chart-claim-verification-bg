# Chart Claim Verification

## Project idea

In this project, I will study whether a neural network can compare a statistical chart with a short text claim.

The model will classify each claim as:

- supported
- refuted
- not enough information

## Research question

Can a model that uses both chart images and text perform better than models that use only one type of input?

## First limits

The first version will use simple line charts and short claims.

The project will start with basic models before a multimodal model is created.
